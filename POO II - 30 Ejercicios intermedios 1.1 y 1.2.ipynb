### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
producto = input("producto:")
cantidad = int(input("cantidad:"))
precio = float(input("precio:"))

subtotal = cantidad * precio
descuento = subtotal * 0.10
iva = (subtotal - descuento) * 0.15
total = subtotal - descuento + iva

print(f"Producto:  {producto}")
print(f"Subtotal:  {subtotal:>10.2f}")
print(f"Descuento: {descuento:>10.2f}")
print(f"IVA:       {iva:>10.2f}")
print(f"TOTAL:     {total:>10.2f}")
````

### Ejercicio 16 · Limpiar un registro ★★

Un sistema exportó este registro con espacios y mayúsculas desordenadas:
`registro = "  0104567896 ,  pérez  ANA ,  ANA.PEREZ@MAIL.COM ,  150.50  "`

### Ejercicio 26 · Dinero exacto con Decimal ★★

1. Muestre el resultado de `0.1 + 0.2` y compruebe si es igual a `0.3`. ¿Qué ocurre?
2. Repita la operación con `Decimal("0.1") + Decimal("0.2")`.
3. Calcule el interés de $1 234.56 al 3.75 % con `Decimal` y redondéelo a centavos con `quantize(Decimal("0.01"))`.
4. Explique en un comentario por qué en un sistema bancario se prefiere `Decimal`.

````python
from decimal import Decimal

suma_float = 0.1 + 0.2
print("0.1 + 0.2 =", suma_float)
print("Es igual a 0.3: ", suma_float == 0.3)

suma_decimal = Decimal("0.1") + Decimal("0.2")
print("Decimal 0.1 + 0.2 =", suma_decimal)
print("Es igual a 0.3: ", suma_decimal == Decimal("0.3"))

capital = Decimal("1234.56")
tasa = Decimal("3.75") / Decimal("100")
interes = (capital * tasa).quantize(Decimal("0.01"))
print("Interés:", interes)
````

Con `correo = "  Ana.Perez@TecAzuay.edu.EC "`:

2. Separe el usuario y el dominio con `split("@")`.

### Ejercicio 29 · Copia o referencia ★★

Ejecute y explique en comentarios cada resultado:
```python
original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()
alias["saldo"] = 500
copia["saldo"] = 999
```
1. Muestre `original`, `alias` y `copia`.
2. Compruebe con `is` cuáles son el mismo objeto.
3. Compare con `id()` las direcciones de memoria.

````python
original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()
alias["saldo"] = 500
copia["saldo"] = 999

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)

print("original is alias:", original is alias)
print("original is copia:", original is copia)

print("id(original):", id(original))
print("id(alias):   ", id(alias))
print("id(copia):   ", id(copia))
````

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"

partes = linea.split("|")
tipo1, saldo1 = partes[2].split("=")
tipo2, saldo2 = partes[3].split("=")
diccionario = {
    "nombre": partes[0].title(),
    "cedula": partes[1], 
    "cuentas": {
        tipo1: float(saldo1),
        tipo2: float(saldo2)
    }
}

saldo_total = sum(diccionario["cuentas"].values())
tipos_cuenta = ", ".join(diccionario["cuentas"].keys())

print(f"Socio: {diccionario['nombre']} ({diccionario['cedula']})")
print(f"Tipos de cuenta: {tipos_cuenta}")
print(f"Saldo total: ${saldo_total:,.2f}")
````
