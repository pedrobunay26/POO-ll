11:  Pida números enteros con input() hasta que el usuario escriba fin. Guárdelos en una lista y luego, con comprensiones, cree una lista de pares y otra de impares. Muestre ambas listas, ordenadas, y la suma de cada una.

```python
numeros = []
while True:
        numero = input("Ingrese un número o 'fin' para finalizar: ")
        if numero == "fin":
            break
        numeros.append(int(numero))

lista_pares = []
lista_impares = []

for i in numeros:
    if i % 2 == 0:
        lista_pares.append(i)
    else:
        lista_impares.append(i)
        
print("\n Lista de pares:", sorted(lista_pares), "Su suma es:", sum(lista_pares))
print(" Lista de impares:", sorted(lista_impares), "Su suma es:", sum(lista_impares))
```
